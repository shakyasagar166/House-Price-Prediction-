# 01. Exploratory Data Analysis (EDA) - Real Estate Housing Market
**Objective**: Analyze residential property attributes, price distributions, neighborhood valuation spreads, living area sizes, and luxury amenity impacts.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

### 1. Load Raw Housing Dataset


```python
df_raw = pd.read_csv('../data/raw/dataset.csv')
print(f"Housing Catalog: {len(df_raw)} properties, {df_raw.shape[1]} attributes")
df_raw.head()
```

### 2. Dataset Schema & Missing Attributes Overview


```python
df_raw.info()
```

```python
# Summary statistics of key numeric dimensions
df_raw.describe()
```

### 3. Property Valuation Distribution


```python
plt.figure(figsize=(8, 4.5))
sns.histplot(df_raw['price'] / 1000, bins=25, kde=True, color='#0284c7')
plt.title('Property Valuation Distribution ($ in Thousands)')
plt.xlabel('Price ($K)')
plt.ylabel('Count')
plt.show()
```

### 4. Valuation by Neighborhood & Property Type


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

order = df_raw.groupby('neighborhood')['price'].median().sort_values(ascending=False).index
sns.barplot(data=df_raw, x='price', y='neighborhood', order=order, palette='Blues_r', ax=axes[0])
axes[0].set_title('Median Home Price by Neighborhood ($)')

sns.boxplot(data=df_raw, x='property_type', y=df_raw['price']/1000, palette='Set2', ax=axes[1])
axes[1].set_title('Valuation Spread by Property Type ($K)')
axes[1].tick_params(axis='x', rotation=25)

plt.tight_layout()
plt.show()
```

### Key EDA Findings:
1. **Neighborhood Premium**: Downtown, Harbor View, and Lakeside command significant valuation premiums.
2. **Living Area**: Strong positive linear correlation between square footage and property price.
3. **Data Quality Note**: Fields like `square_feet` and `lot_size_sqft` contain whitespace strings requiring imputation in Notebook 02.
