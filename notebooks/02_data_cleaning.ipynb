# 02. Data Cleaning & Value Imputation Pipeline
**Objective**: Convert stringified numeric columns, impute missing square footage and lot sizes with medians, standardize boolean flags, and produce cleaned dataset.


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/dataset.csv')
print(f"Raw shape: {df.shape}")
```

### 1. Diagnosing Missing Values & String Formatting


```python
numeric_cols = [
    'square_feet', 'bedrooms', 'bathrooms', 'year_built',
    'lot_size_sqft', 'garage_spaces', 'floors',
    'distance_to_city_center_km', 'school_rating', 'price'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col].astype(str).str.strip(), errors='coerce')

print("Missing values per column after numeric coercion:")
print(df[numeric_cols].isnull().sum())
```

### 2. Robust Median Imputation


```python
# Impute missing square feet, lot sizes, and distance with medians
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print(f"Remaining null values: {df[numeric_cols].isnull().sum().sum()}")
```

### 3. Binary Amenity Standardization


```python
for col in ['has_pool', 'has_garden']:
    df[col] = df[col].apply(lambda x: 1 if str(x).strip().lower() == 'yes' else 0)

df[['square_feet', 'lot_size_sqft', 'has_pool', 'has_garden', 'price']].head()
```

### 4. Export Cleaned Dataset


```python
clean_path = '../data/processed/cleaned_data.csv'
df.to_csv(clean_path, index=False)
print(f"Successfully cleaned and exported {len(df)} properties to: {clean_path}")
```

### Data Cleaning Summary:
- Cleaned string whitespace issues in square footage, lot size, and distance metrics.
- Applied median imputation preserving real estate distribution tails.
- Encoded binary amenities for downstream regression modeling.
